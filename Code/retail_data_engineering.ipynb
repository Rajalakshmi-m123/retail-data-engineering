## 5. Data Cleaning & Transformation

```python
retail_clean = retail_raw.copy()

print(retail_clean.shape)
```

```python
sorted(retail_clean['category'].unique())
```

```python
category_mapping = {
    'ELEC': 'Electronics',
    'electronics': 'Electronics',

    'FURN': 'Furniture',
    'furniture': 'Furniture',

    'CLOTH': 'Clothing',
    'clothing': 'Clothing',

    'HOME': 'Home Appliances',
    'home appliances': 'Home Appliances'
}

retail_clean['category'] = retail_clean['category'].replace(category_mapping)
```

```python
sorted(retail_clean['category'].unique())
```

```python
sorted(retail_clean['product_name'].unique())
```

```python
retail_clean['product_name'] = retail_clean['product_name'].str.title()
```

```python
sorted(retail_clean['product_name'].unique())
```

```python
sorted(product['product_name'].unique())
```

```python
retail_clean['price'].isnull().sum()
```

```python
retail_clean[retail_clean['price'].isnull()].head()
```

```python
retail_clean[retail_clean['price'].isnull()][['product_id','product_name','price']].head(10)
```

```python
price_mapping = product.set_index('product_id')['price']

price_mapping
```

```python
retail_clean['price'] = retail_clean['price'].fillna(
    retail_clean['product_id'].map(price_mapping)
)
```

```python
retail_clean['price'].isnull().sum()
```

Missing prices were not imputed using statistical measures such as mean or median. Instead, missing values were populated using the standardized product price available in the product reference table (product_details), ensuring business consistency and data accuracy.

## 6. PII Masking

```python
retail_clean[['email','phone']].head()
```

```python
def mask_email(email):
    username, domain = email.split('@')
    return username[0] + '*' * (len(username)-1) + '@' + domain

retail_clean['email_masked'] = retail_clean['email'].apply(mask_email)
```

```python
retail_clean[['email','email_masked']].head()
```

```python
def mask_phone(phone):
    phone = str(phone)
    return phone[:2] + '*' * 6 + phone[-2:]

retail_clean['phone_masked'] = retail_clean['phone'].apply(mask_phone)
```

```python
retail_clean[['phone','phone_masked']].head()
```

```python
retail_clean.drop(
    columns=['email','phone'],
    inplace=True
)
```

```python
retail_clean.columns
```

## 7. Product Dimension Enrichment

```python
retail_enriched = retail_clean.merge(
    product,
    on='product_id',
    how='left',
    suffixes=('_txn', '_master')
)

retail_enriched.shape
```

```python
retail_enriched.head()
```

## 8. KPI Generation

```python
retail_enriched['revenue'] = (
    retail_enriched['price_txn']
    * retail_enriched['quantity']
    * (1 - retail_enriched['discount'])
)
```

```python
retail_enriched[['price_txn','quantity','discount','revenue']].head()
```

KPI 1: Total Revenue

```python
total_revenue = retail_enriched['revenue'].sum()

print(f"Total Revenue: ₹{total_revenue:,.2f}")
```

KPI 2: Revenue by Category

```python
revenue_by_category = (
    retail_enriched
    .groupby('category_txn')['revenue']
    .sum()
    .sort_values(ascending=False)
)

revenue_by_category
```

KPI 3: Revenue by City

```python
revenue_by_city = (
    retail_enriched
    .groupby('city')['revenue']
    .sum()
    .sort_values(ascending=False)
)

revenue_by_city
```

KPI 4: Top Selling Products

```python
top_products = (
    retail_enriched
    .groupby('product_name_txn')['revenue']
    .sum()
    .sort_values(ascending=False)
)

top_products.head(10)
```

KPI 5: Revenue by Payment Method

```python
revenue_by_payment = (
    retail_enriched
    .groupby('payment_method')['revenue']
    .sum()
    .sort_values(ascending=False)
)

revenue_by_payment
```

KPI 6: Monthly Revenue Trend

```python
monthly_revenue = (
    retail_enriched
    .groupby(retail_enriched['transaction_date'].dt.to_period('M'))['revenue']
    .sum()
)

monthly_revenue
```
